```python
import pandas as pd

# 1. Load the finalized dataset
processed_path = "../data/processed/macro_sector_features.csv.gz"
print(f"Loading processed data from {processed_path}...\n")
df_final = pd.read_csv(processed_path, compression='gzip')

# 2. Shape and Data Types
print("--- DATASET SHAPE ---")
print(f"Rows: {df_final.shape[0]:,}")
print(f"Columns: {df_final.shape[1]}")

print("\n--- MISSING VALUES ---")
# Calculate percentage of missing data per column
missing_pct = (df_final.isnull().sum() / len(df_final)) * 100
print(missing_pct.round(2).astype(str) + '%')

print("\n--- DATA SNEAK PEEK ---")
# Display a random sample of 5 rows
display(df_final.sample(5))
```
